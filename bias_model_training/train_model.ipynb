
### Pearson Correlation (Counts) Summary: Peaks vs. Non-Peaks
#### Using the 0.6 bias model based on these data

| Model Name           | Pearson Correlation (Non-peaks) | Pearson Correlation (Peaks) |
| :------------------- | :-----------------------------: | :-------------------------: |
| bias_model_f0_b0.3|             -0.459              |           -0.374            |
| bias_model_f0_b0.4|             -0.355              |           -0.486            |
| bias_model_f0_b0.5|             0.254           |           -0.538            |
| **bias_model_f0_b0.6**|             0.454           |           -0.405            |
| bias_model_f0_b0.7|             0.437           |           -0.472            |
| bias_model_f0_b0.8|             0.373           |           -0.477            |



```python
import matplotlib.pyplot as plt
import numpy as np

# Data from the bias model evaluations
model_names = ['bias_model_f0_b0.3', 'bias_model_f0_b0.4', 'bias_model_f0_b0.5', 
               'bias_model_f0_b0.6', 'bias_model_f0_b0.7', 'bias_model_f0_b0.8']
non_peaks_correlations = [-0.459, -0.355, 0.254, 0.454, 0.437, 0.373]
peaks_correlations = [-0.374, -0.486, -0.538, -0.405, -0.472, -0.477]

# Create bar plot
fig, ax = plt.subplots(figsize=(12, 6))

x = np.arange(len(model_names))
width = 0.35

# Create bars
bars1 = ax.bar(x - width/2, non_peaks_correlations, width, label='Non-peaks', alpha=0.8)
bars2 = ax.bar(x + width/2, peaks_correlations, width, label='Peaks', alpha=0.8)

# Add labels over each bar
for i, (bar1, bar2) in enumerate(zip(bars1, bars2)):
    # Label for non-peaks bars
    height1 = bar1.get_height()
    ax.text(bar1.get_x() + bar1.get_width()/2., height1 + (0.02 if height1 >= 0 else -0.05),
            f'{non_peaks_correlations[i]:.3f}', ha='center', va='bottom' if height1 >= 0 else 'top', fontsize=9)
    
    # Label for peaks bars
    height2 = bar2.get_height()
    ax.text(bar2.get_x() + bar2.get_width()/2., height2 + (0.02 if height2 >= 0 else -0.05),
            f'{peaks_correlations[i]:.3f}', ha='center', va='bottom' if height2 >= 0 else 'top', fontsize=9)

# Customize the plot
ax.set_xlabel('Model Name')
ax.set_ylabel('Pearson Correlation')
ax.set_title('Pearson Correlation: Peaks vs. Non-Peaks Across Bias Models')
ax.set_xticks(x)
ax.set_xticklabels(model_names, rotation=45, ha='right')
ax.legend()
ax.grid(axis='y', alpha=0.3)

# Add horizontal line at y=0
ax.axhline(y=0, color='black', linestyle='-', alpha=0.3)

# Highlight the best model (bias_model_f0_b0.6)
best_model_idx = 3
bars1[best_model_idx].set_edgecolor('red')
bars1[best_model_idx].set_linewidth(2)
bars2[best_model_idx].set_edgecolor('red')
bars2[best_model_idx].set_linewidth(2)

plt.tight_layout()
plt.show()

# Print summary statistics
print("Summary Statistics:")
print("==================")
for i, model in enumerate(model_names):
    print(f"{model}: Non-peaks={non_peaks_correlations[i]:.3f}, Peaks={peaks_correlations[i]:.3f}")
```
