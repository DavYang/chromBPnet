```python
#Cycle through each bias threshold and generate a table of the model count metrics.

import json
import os
import pandas as pd
from pathlib import Path

# Set the directory to search for model results
results_directory = "/gs/gsfs0/shared-lab/greally-lab/David/chromBPnet_analysis/bias_model_training/outputs/pbmc_merged"

# Define the results directory pattern
results_pattern = "bias_model_f0_b*"
evaluation_subdir = "evaluation"
metrics_filename = "pbmc_merged_bias_metrics.json"

# Find all model directories in the specified directory
model_dirs = sorted([d for d in os.listdir(results_directory) if d.startswith('bias_model_f0_b')])

print(f"Found {len(model_dirs)} model directories in {results_directory}")

# Initialize lists to store data for table
table_data = []

# Cycle through each model directory
for model_dir in model_dirs:
    evaluation_path = os.path.join(results_directory, model_dir, evaluation_subdir)
    metrics_file = os.path.join(evaluation_path, metrics_filename)
    
    if os.path.exists(metrics_file):
        try:
            with open(metrics_file, 'r') as f:
                data = json.load(f)
            
            # Extract pearsonr values
            peaks_pearson = data['counts_metrics']['peaks']['pearsonr']
            nonpeaks_pearson = data['counts_metrics']['nonpeaks']['pearsonr']
            
            # Extract bias threshold from model directory name
            bias_threshold = model_dir.split('_b')[-1]
            
            # Add to table data
            table_data.append({
                'Model': model_dir,
                'Bias_Threshold': bias_threshold,
                'Peaks_Pearsonr': peaks_pearson,
                'NonPeaks_Pearsonr': nonpeaks_pearson
            })
            
        except (json.JSONDecodeError, KeyError) as e:
            print(f"Error parsing {metrics_file}: {e}")
    else:
        print(f"Warning: {metrics_file} not found")

# Create DataFrame and display table
df = pd.DataFrame(table_data)

if not df.empty:
    print("\nPearson Correlation Results Table:")
    print("=" * 60)
    print(df.to_string(index=False, float_format='%.4f'))
    
    # Find best performing models
    best_peaks_idx = df['Peaks_Pearsonr'].idxmax()
    best_nonpeaks_idx = df['NonPeaks_Pearsonr'].idxmax()
    
    print(f"\nBest performing models:")
    print(f"Peaks: {df.iloc[best_peaks_idx]['Model']} (r={df.iloc[best_peaks_idx]['Peaks_Pearsonr']:.4f})")
    print(f"Non-peaks: {df.iloc[best_nonpeaks_idx]['Model']} (r={df.iloc[best_nonpeaks_idx]['NonPeaks_Pearsonr']:.4f})")
else:
    print("No valid metrics data found")
```
