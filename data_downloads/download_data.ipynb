# Downloading ATAC-seq data from SRA
The goal is to download the FASTQ files for the ATAC-seq data from SRA which represent major blood cell types for analysis using ChromBPnet.
* Neutrophils: 40% to 60%
* Lymphocytes: 20% to 40%
* Monocytes: 2% to 8%
* Eosinophils: 1% to 4%
* Basophils: 0.5% to 1%
* Band (young neutrophil): 0% to 3% 
[source](https://www.ucsfhealth.org/medical-tests/blood-differential-test)


### Neutrophil data
* celltype: neutrophil
* sample info: 30 year old female
* Sequencing platform: Illumina HiSeq X Ten, paired end sequencing
* [source paper](https://pmc.ncbi.nlm.nih.gov/articles/PMC8321655/#s26)

The following commands are executed inside a bash script to download the SRA files and convert them to FASTQ files (and compress them)

```python
# Copy and paste into the terminal while cd'ed into the data_downloads directory
"The combination of prefetch + fasterq-dump is the fastest way to extract FASTQ-files from SRA-accessions. The prefetch tool downloads all necessary files to your computer. The prefetch - tool can be invoked multiple times if the download did not succeed. It will not start from the beginning every time; instead, it will pick up from where the last invocation failed."

prefetch SRR11909927 # downloads the SRA file for this sample
vdb-validate SRR11909927 # validates the SRA file
fasterq-dump SRR11909927 --outdir ./SRR11909927/ # downloads the FASTQ files for this sample
gzip SRR11909927/*.fastq # compress the FASTQ files
```
