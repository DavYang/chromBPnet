This notebook is used to generate the data used to train the Tn5 bias model. The following code will execute 2 steps: 
1. Divide each sample genome into 3 sets of chromosomes: test, validation, and training. 
2. Generate non-peak regions for each fold which are matched for GC content with respect to the peak regions. 

* NOTE: In order to run chrombpnet, we will use singularity rather than trying to download it using pip. The singularity image should already be downloaded in the bash directory and can be found in the script.
